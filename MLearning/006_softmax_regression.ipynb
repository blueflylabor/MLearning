```python
import torch
```

```python
num_inputs, num_outputs, num_samples = 784, 3, 1000
X = torch.randn(num_samples, num_inputs)
y = torch.randint(0, num_outputs, (num_samples, ))
X, y
```

```python
W = torch.normal(0, 0.01, size=(num_inputs, num_outputs), requires_grad=True)
b = torch.zeros(num_outputs, requires_grad=True)
W, b
```

```python
def softmax(O):
    O_max = torch.max(O, dim=1, keepdim=True).values
    exp_O = torch.exp(O - O_max)
    partition = exp_O.sum(dim=1, keepdim=True)
    return exp_O / partition
```

```python
def reg(X):
    return softmax(X @ W + b)
```

```python
def cross_entropy(y_, y):
    return - torch.log(y_[range(len(y_)), y] + 1e-7)
```

```python
lr, batch_size, eps = 0.1, 32, 1000
```

```python
for ep in range(eps):
    inds = torch.randperm(eps)
    for i in range(0, num_samples, batch_size):
        batch_idx = inds[i : i + batch_size]
        X_batch = X[batch_idx]
        y_batch = y[batch_idx]

        y_ = reg(X_batch)
        loss = cross_entropy(y_, y_batch).mean()
        loss.backward()

        with torch.no_grad():
            W -= lr * W.grad
            b -= lr * b.grad

            W.grad.zero_()
            b.grad.zero_()

    with torch.no_grad():
        all_y_hat = reg(X)
        preds = all_y_hat.argmax(dim=1)
        acc = (preds == y).float().mean()
        total_loss = cross_entropy(all_y_hat, y).mean()
        print(f"Ep {ep+1:2d} | Loss:{total_loss.item():.4f} | Acc:{acc.item() * 100:.2f}%")
```

```python
import numpy as np

# 1. 设定随机种子与生成合成数据
np.random.seed(42)
N = 100  # 样本数
D = 3    # 特征数

# 真实参数 (目标: 算法应该学习出接近这些值的 W 和 b)
true_W = np.array([[2.0], [-1.5], [0.5]]) # (3, 1)
true_b = 3.0                              # 标量

# 生成输入 X (100, 3) 和 带有噪声的标签 Y (100, 1)
X = np.random.randn(N, D)
Y = X @ true_W + true_b + np.random.randn(N, 1) * 0.1

# 2. 初始化模型参数 (W, b)
W = np.random.randn(D, 1) * 0.01  # (3, 1)
b = np.zeros((1, 1))              # (1, 1)

# 3. 超参数设置
learning_rate = 0.1
epochs = 200

print("--- 开始训练 ---")
for epoch in range(epochs):
    # ====================================================
    # A. 前向传播 (Forward Pass)
    # ====================================================
    # 1) 矩阵相乘 + 广播加偏置: (100, 3) @ (3, 1) + (1, 1) -> (100, 1)
    Z = np.dot(X, W) + b
    
    # 2) 计算 MSE Loss: 标量
    loss = np.mean((Z - Y) ** 2) / 2.0
    
    # ====================================================
    # B. 反向传播 (Backward Pass - 链式法则传递梯度)
    # ====================================================
    # Step 1: dL / dZ (损失关于预测值 Z 的梯度)
    # 数学公式: dZ = (Z - Y) / N
    dZ = (Z - Y) / N   # (100, 1)
    
    # Step 2: dL / dW (上游梯度 dZ 乘以 dZ/dW，即 X^T)
    # 数学公式: dW = X^T @ dZ
    dW = np.dot(X.T, dZ) # (3, 100) @ (100, 1) -> (3, 1)
    
    # Step 3: dL / db (偏置项沿 Batch 维度累加)
    # 数学公式: db = sum(dZ)
    db = np.sum(dZ, axis=0, keepdims=True) # (1, 1)
    
    # ====================================================
    # C. 参数更新 (Gradient Descent)
    # ====================================================
    W -= learning_rate * dW
    b -= learning_rate * db
    
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f"Epoch {epoch+1:2d} | Loss: {loss:.6f}")

print("\n--- 训练结束，拟合结果对比 ---")
print("真实 W:\n", true_W.ravel())
print("学到的 W:\n", W.ravel())
print(f"真实 b: {true_b:.2f} | 学到的 b: {b.item():.2f}")
```
