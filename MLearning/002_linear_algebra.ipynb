```python
import torch

A = torch.tensor([[1.0, 2.0],
                  [3.0, 4.0]])

# 方式 1：使用 PyTorch 内置函数
print("F 范数:", torch.norm(A)) # 结果: sqrt(1^2 + 2^2 + 3^2 + 4^2) = sqrt(30) ≈ 5.4772

# 方式 2：手动计算验证
print("验证:", torch.sqrt(torch.sum(A**2))) # 5.4772
```

```python
import torch

# 假设有一个权重向量 w
w = torch.tensor([3.0, -4.0])

# 1. L2 范数 (默认)
l2 = torch.norm(w, p=2)  # 或 torch.linalg.vector_norm(w, ord=2)
print("L2 范数:", l2)     # sqrt(3^2 + (-4)^2) = 5.0

# 2. L1 范数
l1 = torch.norm(w, p=1)  # |3| + |-4|
print("L1 范数:", l1)     # 7.0

# 3. L_infinity 范数
l_inf = torch.norm(w, p=float('inf'))
print("L_inf 范数:", l_inf) # max(3, 4) = 4.0
```
