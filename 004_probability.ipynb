```python
import torch
from torch.distributions import Multinomial
```

```python
prob = torch.ones([6]) / 6
samples = Multinomial(10, prob).sample()
samples
```

```python
counts = Multinomial(10000, prob).sample()
counts / 10000
```

```python
x = torch.tensor([1, 2, 3, 5.])
p = torch.tensor([0.1, 0.2, 0.4, 0.6])

exp = torch.sum(x * p)
exp.item()
```

```python
var = torch.sum(((x - exp) ** 2) * p)
var.item()
```
